```python
import os
import random
import matplotlib.pyplot as plt
from PIL import Image
```

```python
DATASET_PATH = "../data/faces"

original_path = os.path.join(DATASET_PATH, "original")
deepfake_path = os.path.join(DATASET_PATH, "deepfakes")

print("Original path:", original_path)
print("Deepfake path:", deepfake_path)
```

```python
def get_random_images(folder_path, count=6):
    images = []

    for root, dirs, files in os.walk(folder_path):
        for file in files:
            if file.endswith(".jpg"):
                images.append(os.path.join(root, file))

    return random.sample(images, count)
```

```python
original_samples = get_random_images(original_path, 6)
deepfake_samples = get_random_images(deepfake_path, 6)

print("Original samples:")
for img in original_samples:
    print(img)

print("\nDeepfake samples:")
for img in deepfake_samples:
    print(img)
```

```python
id="1"
def show_images(image_paths, title):
    plt.figure(figsize=(12, 6))

    for i, img_path in enumerate(image_paths):
        img = Image.open(img_path)

        plt.subplot(2, 3, i + 1)
        plt.imshow(img)
        plt.axis("off")

    plt.suptitle(title, fontsize=16)
    plt.show()
```

```python
show_images(original_samples, "Original Face Samples")
```

```python
show_images(deepfake_samples, "Deepfake Face Samples")
```
